#1. Сбор требований

Для формализации сбора требований к разрабатываемому программному обеспечению с исользованиме гибкой (англ. Agile) методологии будет применем метод историй пользователей (англ. User Story), в котором требование представляется в виде краткого текста, имеющему такую структуру, из которой можно выделить смысловые фрагменты для дальнейшего использования при постановке задачи зарработчикам и ее приоритизации в бэклоге продукта (англ. Product Backlog).

**Шаблон User Story: «Как (роль), я хочу (действие/ возможность), чтобы (ценность/ результат)»**

Задача: предложить лингвистическую конструкцию User Story, которая будет содержать необходимый набор оценок для дальнейшей приоритизации как в четком, так и в нечетком виде для проведения дальнейших расчетов, сравнительного анализа и выводов о целесообразности использования нечетких оценок для управления требованиям (в частности, на этапе сбора и приоритизации). Т.е. суть подхода заключается в следующем: с одной стороны, заказчик отвечает на вопросы обычным разговорным языком, с другой – эксперт из команды разработки фиксирует необходимые оценки параметров по лингвистическим шкалам для дальнейшей приоритизации задачи в бэклоге продукта.

**Возможные критерии постановки задачи:**

**Сложность (англ. Complexity)** – насколько специфичными должны быть знания и уровень квалификации персонала;

**Важность, ценность (англ. Importance/ Value)** – польза для бизнеса/ пользователя;

**Срочность (англ. Urgency/ Time criticality)** – насколько быстро нужно реализовать задачу;

**Трудоемкость (англ. Effort)** – сколько усилий нужно, оценка масштабности задачи;

**Риски (англ. Risks)** – насколько опасно, если не сделать, вероятность проблемы и возможный ущерб.


**Например, для приоритизации по методу RICE необходимо оценить следующие характеристики продукта:**

**Охват (англ. Reach)**, единицы измерения - количество человек, ед.;

**Уверенность (англ. Confidence)**,  коэффициент, шкала от 0,5 до 1;

**Трудозатраты (англ. Effort)**, единицы измерения - количество недель, ед.

Четкая модель приоритизации по методу RICE

```python
def RICE (R, I, C, E):
    RICE = round(R * I* C / E, 2)
    return RICE
```

```python
import pandas as pd

def rice_score(reach: float, impact: float, confidence: float, effort: float) -> float:
    if effort == 0:
        raise ValueError("Effort (E) не может быть 0")
    return round(reach * impact * confidence / effort, 2)


features = [
    {"Feature": "Feature A", "Reach (R)": 10000, "Impact (I)": 1, "Confidence (C)": 0.6, "Effort (E)": 10},
    {"Feature": "Feature B", "Reach (R)": 100,   "Impact (I)": 3, "Confidence (C)": 0.9, "Effort (E)": 5},
    {"Feature": "Feature C", "Reach (R)": 1000,  "Impact (I)": 2, "Confidence (C)": 0.8, "Effort (E)": 7},
]

df = pd.DataFrame(features)

# считаем RICE
df["RICE Score"] = df.apply(
    lambda r: rice_score(r["Reach (R)"], r["Impact (I)"], r["Confidence (C)"], r["Effort (E)"]),
    axis=1
)

# сортируем по убыванию RICE и добавляем ранг
df = df.sort_values("RICE Score", ascending=False).reset_index(drop=True)
df.insert(0, "Rank", df.index + 1)

# вывод
df_pretty = df.copy()
df_pretty["Reach (R)"] = df_pretty["Reach (R)"].map(lambda x: f"{int(x):,}".replace(",", " "))
df_pretty["RICE Score"] = df_pretty["RICE Score"].map(lambda x: f"{x:,.2f}".replace(",", " "))

print(df_pretty.to_string(index=False))
```

```python
from pathlib import Path
import re
import pandas as pd

def _normalize_col(s: str) -> str:
    return " ".join(str(s).strip().lower().split())

def _detect_columns(columns):
    cols_norm = {_normalize_col(c): c for c in columns}
    originals = list(columns)

    synonyms = {
        "Feature": ["feature", "name", "название", "название фичи", "фича", "задача"],
        "R": ["r", "reach", "охват", "reach (r)", "охват (r)"],
        "I": ["i", "impact", "влияние", "impact (i)", "влияние (i)"],
        "C": ["c", "confidence", "уверенность", "confidence (c)", "уверенность (c)"],
        "E": ["e", "effort", "трудозатраты", "трудоемкость", "усилия", "effort (e)", "трудозатраты (e)"],
    }

    detected = {}
    for canonical, syns in synonyms.items():
        found = None

        # 1) точное совпадение по словарю синонимов
        for syn in syns:
            syn_norm = _normalize_col(syn)
            if syn_norm in cols_norm:
                found = cols_norm[syn_norm]
                break

        # 2) поиск по шаблону "(r)" "(i)" "(c)" "(e)"
        if found is None and canonical in ["R", "I", "C", "E"]:
            key = canonical.lower()
            for c in originals:
                cn = _normalize_col(c)
                if re.search(rf"\({key}\)", cn):
                    found = c
                    break

        # 3) поиск по ключевым словам (на всякий случай)
        if found is None:
            keywords = {
                "Feature": ["feature", "назван", "задач", "фич"],
                "R": ["reach", "охват"],
                "I": ["impact", "влияни"],
                "C": ["confidence", "уверенн"],
                "E": ["effort", "трудозат", "трудоемк", "усили"],
            }[canonical]
            for c in originals:
                cn = _normalize_col(c)
                if any(k in cn for k in keywords):
                    found = c
                    break

        if found is None:
            raise ValueError(f"Не найдена колонка для '{canonical}'. Доступные колонки: {list(columns)}")
        detected[canonical] = found

    return detected

def compute_rice_from_file(input_path: str, output_path: str | None = None, sheet_name: str | int | None = 0) -> pd.DataFrame:
    input_path = Path(input_path)
    if not input_path.exists():
        raise FileNotFoundError(f"Файл не найден: {input_path}")

    # 1) читаем файл
    suffix = input_path.suffix.lower()
    if suffix in [".xlsx", ".xls"]:
        df = pd.read_excel(input_path, sheet_name=sheet_name)  # 0 = первый лист
    elif suffix == ".csv":
        df = pd.read_csv(input_path)
    else:
        raise ValueError("Поддерживаются только .xlsx/.xls и .csv")

    # 2) определяем и нормализуем названия колонок
    colmap = _detect_columns(df.columns)
    df = df.rename(columns={
        colmap["Feature"]: "Feature",
        colmap["R"]: "R",
        colmap["I"]: "I",
        colmap["C"]: "C",
        colmap["E"]: "E",
    })

    # 3) приводим типы + поддержка десятичной запятой
    for col in ["R", "I", "C", "E"]:
        df[col] = df[col].astype(str).str.replace(",", ".", regex=False)
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # 4) валидации
    bad = df[df[["Feature", "R", "I", "C", "E"]].isna().any(axis=1)]
    if not bad.empty:
        raise ValueError("Есть строки с пустыми/некорректными значениями:\n" + bad.to_string(index=False))
    if (df["E"] == 0).any():
        raise ValueError("Есть строки с Effort (E) = 0 — деление невозможно.")

    # (опционально) если confidence задан в процентах 60 вместо 0.6
    if (df["C"] > 1).any() and (df["C"] <= 100).all():
        df["C"] = df["C"] / 100.0

    # 5) расчет + ранжирование
    df["RICE"] = (df["R"] * df["I"] * df["C"] / df["E"]).round(2)
    df = df.sort_values("RICE", ascending=False).reset_index(drop=True)
    df.insert(0, "Rank", df.index + 1)

    # 6) сохранение
    if output_path is None:
        output_path = str(input_path.with_name(input_path.stem + "_OUT" + input_path.suffix))

    out = Path(output_path)
    if out.suffix.lower() in [".xlsx", ".xls"]:
        df.to_excel(out, index=False)
    elif out.suffix.lower() == ".csv":
        df.to_csv(out, index=False)
    else:
        out = out.with_suffix(".xlsx")
        df.to_excel(out, index=False)

    print(f"Результат сохранён в: {out}")
    return df
```

**2.2. Нечеткий подход**

```python
# Функция произведения двуж нечетких трапециевидных чисел
def multiply_intuitionistic_trapezoidal_fuzzy_numbers(X, Y):
    # Распаковка параметров интервальных трапециевидных чисел 2 типа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = X
    (y1, y2, y3, y4), (y5, y6, y7, y8) = Y

    # Вычисление координат для произведения
    z1 = x2 * y1 - x2 * y2 + x1 * y2
    z2 = x2 * y2
    z3 = x3 * y3
    z4 = x3 * y4 - x3 * y3 + y3 * x4

    z5 = x6 * y5 - x6 * y6 + x5 * y6
    z6 = x6 * y6
    z7 = x7 * y7
    z8 = x7 * y8 - x7 * y7 + y7 * x8

    return ((z1, z2, z3, z4), (z5, z6, z7, z8))
```

```python
# Функция вычмсления произведения нескольких нечетких чисел, для данной модели - показателей R, I, C в нечетком виде
def calculate_RIC_multiply(*params):
    RIC_multiply = params[0]
    for param in params[1:]:
        RIC_multiply = multiply_intuitionistic_trapezoidal_fuzzy_numbers (RIC_multiply, param)
    return RIC_multiply
```

```python
# Функция округления параметров трапециевидного нечеткого числа (интуиционистского или интервального 2 типа)
def round_fuzzy_number(fuzzy_number):
    (x1, x2, x3, x4), (x5, x6, x7, x8) = fuzzy_number
    return ((round(x1, 3), round(x2, 3), round(x3, 3), round(x4, 3)),
            (round(x5, 3), round(x6, 3), round(x7, 3), round(x8, 3)))
```

```python
# Функция деления отдного нечеткого трапециевичного числа на другое
def divide_intuitionistic_trapezoidal_fuzzy_numbers(X, Y):
    # Распаковка параметров интервальных трапециевидных чисел 2 типа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = X
    (y1, y2, y3, y4), (y5, y6, y7, y8) = Y

    # Вычисление координат для частного от деления
    z1 = (x2 * y3 - x2 * y4 + x1 * y3) / ((y3)**2)
    z2 = x2 / y3
    z3 = x3 / y2
    z4 = (x3 * y2 - x3 * y1 + x4 * y2) / ((y2)**2)

    z5 = (x6 * y7 - x6 * y8 + x5 * y7) / ((y7)**2)
    z6 = x6 / y7
    z7 = x7 / y6
    z8 = (x7 * y6 - x7 * y5 + x8 * y6) / ((y6)**2)

    return ((z1, z2, z3, z4), (z5, z6, z7, z8))
```

```python
# Функция для вычисления альфа-среза
def alpha_cut(fuzzy_number, alpha):
    # Распаковка параметров нечеткого числа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = fuzzy_number

    # Вычисление концов верхнего интервала альфа-среза
    upper_start = alpha * x2 + (1 - alpha) * x1
    upper_end = alpha * x3 + (1 - alpha) * x4

    # Вычисление концов нижнего интервала альфа-среза
    lower_start = alpha * x6 + (1 - alpha) * x5
    lower_end = alpha * x7 + (1 - alpha) * x8

    # Округляем значения до 3 знаков после запятой
    upper_start = round(upper_start, 3)
    upper_end = round(upper_end, 3)
    lower_start = round(lower_start, 3)
    lower_end = round(lower_end, 3)

    # Возвращаем альфа-срез в виде нечеткого числа
    return ((upper_start, upper_end), (lower_start, lower_end))
```

```python
# Функция вычисления центроида системы двух точек
def centroid(alpha_cut_result):
    # Распаковка верхнего и нижнего отрезков альфа-среза
    (upper_start, upper_end), (lower_start, lower_end) = alpha_cut_result

    # Расчет среднего значения
    central_upper = round((upper_start + upper_end) / 2, 3)
    central_lower = round((lower_start + lower_end) / 2, 3)

    # Возвращаем результат в нужном формате
    return (central_upper, central_lower)
```

```python
# Функция вычисления среднего арифметического дефаззифицированных значений двух функций принадлежности
def calculate_average(defuzzified_values):
    # Извлекаем значения из кортежа
    defuzzified_upper, defuzzified_lower = defuzzified_values
    # Вычисляем среднее арифметическое
    average = round((defuzzified_upper + defuzzified_lower) / 2, 3)
    return average
```

**Расчет показателя RICE по нечеткой модели для задачи 1**

```python
# Ввод показателей R, I, C, E в нечетком виде.
R = ((9700, 9900, 10100, 10300), (9800, 9900, 10100, 10200))  # Reach (Охват), количество человек
I = ((0.97, 0.99, 1.01, 1.03), (0.98, 0.99, 1.01, 1.02)) # Impact (Влияние), оценка в баллах
C = ((0.58, 0.59, 0.61, 0.62), (0.59, 0.59, 0.61, 0.61)) # Confidence (Уверенность), коэффициент от 0,5 до 1
E = ((9.70, 9.90, 10.10, 10.30), (9.80, 9.90, 10.10, 10.20)) # Effort (Трудозатраты), количество недель

RIC_mult = calculate_RIC_multiply(R, I, C)
# Округляем RIC
RIC_rounded = round_fuzzy_number(RIC_mult)

print(f"При вышеуказанных значениях параметров R, I, C, произведение RIC в нечетком виде принимает значение {RIC_rounded}")
```

```python
RICE_score = divide_intuitionistic_trapezoidal_fuzzy_numbers(RIC_mult, E)
# Округляем RICE
RICE_rounded = round_fuzzy_number(RICE_score)

print(f"При вышеуказанных значениях параметров R, I, C, E показатель RICE в нечетком виде принимает значение {RICE_rounded}")
```

```python
# Вычисление альфа-среза
alpha = 0.8
alpha_cut_result = alpha_cut(RICE_rounded, alpha)
print(f"Альфа-срез при альфа={alpha} для RICE_rounded {RICE_rounded} равен {alpha_cut_result}")
```

```python
# Расчет центроида системы двух точек
central_result = centroid(alpha_cut_result)

print(f"Усредненные значения для RICE_rounded {RICE_rounded} при альфа={alpha} равны {central_result}")
```

# Функции - сновные операции с нечеткими числами

**Умножение двух нечетких трапециевидных чисел 2-го типа**

```python
def multiply_intuitionistic_trapezoidal_fuzzy_numbers(X, Y):
    # Распаковка параметров интервальных трапециевидных чисел 2 типа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = X
    (y1, y2, y3, y4), (y5, y6, y7, y8) = Y

    # Вычисление координат для произведения
    z1 = x2 * y1 - x2 * y2 + x1 * y2
    z2 = x2 * y2
    z3 = x3 * y3
    z4 = x3 * y4 - x3 * y3 + y3 * x4

    z5 = x6 * y5 - x6 * y6 + x5 * y6
    z6 = x6 * y6
    z7 = x7 * y7
    z8 = x7 * y8 - x7 * y7 + y7 * x8

    return ((z1, z2, z3, z4), (z5, z6, z7, z8))
```

**Деление одного нечеткого трапециевидного числа 2-го типа на другое нечеткое трапециевидное число 2-го типа**

```python
def divide_intuitionistic_trapezoidal_fuzzy_numbers(X, Y):
    # Распаковка параметров интервальных трапециевидных чисел 2 типа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = X
    (y1, y2, y3, y4), (y5, y6, y7, y8) = Y

    # Вычисление координат для частного от деления
    z1 = (x2 * y3 - x2 * y4 + x1 * y3) / ((y3)**2)
    z2 = x2 / y3
    z3 = x3 / y2
    z4 = (x3 * y2 - x3 * y1 + x4 * y2) / ((y2)**2)

    z5 = (x6 * y7 - x6 * y8 + x5 * y7) / ((y7)**2)
    z6 = x6 / y7
    z7 = x7 / y6
    z8 = (x7 * y6 - x7 * y5 + x8 * y6) / ((y6)**2)

    return ((z1, z2, z3, z4), (z5, z6, z7, z8))
```

**Сумма двух нечетких трапециевидных чисел 2-го типа**

```python
# Функция сложения двух трапециевидных нечетких чисел (интуиционистских или интервальных 2 типа)
def add_intuitionistic_trapezoidal_fuzzy_numbers(fuzzy_number1, fuzzy_number2):
    (x1, x2, x3, x4), (x5, x6, x7, x8) = fuzzy_number1
    (y1, y2, y3, y4), (y5, y6, y7, y8) = fuzzy_number2
    return ((round(x1 + y1, 3), round(x2 + y2, 3), round(x3 + y3, 3), round(x4 + y4, 3)),
            (round(x5 + y5, 3), round(x6 + y6, 3), round(x7 + y7, 3), round(x8 + y8, 3)))
```

**Округление нечеткого трапециевидного числа 2-го типа**

```python
# Функция округления параметров трапециевидного нечеткого числа (интуиционистского или интервального 2 типа)
def round_fuzzy_number(fuzzy_number):
    (x1, x2, x3, x4), (x5, x6, x7, x8) = fuzzy_number
    return ((round(x1, 3), round(x2, 3), round(x3, 3), round(x4, 3)),
            (round(x5, 3), round(x6, 3), round(x7, 3), round(x8, 3)))
```

**Вычисление границ альфа-среза**

```python
# Функция для вычисления альфа-среза
def alpha_cut(fuzzy_number, alpha):
    # Распаковка параметров нечеткого числа
    (x1, x2, x3, x4), (x5, x6, x7, x8) = fuzzy_number

    # Вычисление концов верхнего интервала альфа-среза
    upper_start = alpha * x2 + (1 - alpha) * x1
    upper_end = alpha * x3 + (1 - alpha) * x4

    # Вычисление концов нижнего интервала альфа-среза
    lower_start = alpha * x6 + (1 - alpha) * x5
    lower_end = alpha * x7 + (1 - alpha) * x8

    # Округляем значения до 3 знаков после запятой
    upper_start = round(upper_start, 3)
    upper_end = round(upper_end, 3)
    lower_start = round(lower_start, 3)
    lower_end = round(lower_end, 3)

    # Возвращаем альфа-срез в виде нечеткого числа
    return ((upper_start, upper_end), (lower_start, lower_end))
```

**Нахождение центроида системы из двух точек**

```python
# Функция вычисления центроида системы двух точек
def centroid(alpha_cut_result):
    # Распаковка верхнего и нижнего отрезков альфа-среза
    (upper_start, upper_end), (lower_start, lower_end) = alpha_cut_result

    # Расчет среднего значения
    central_upper = round((upper_start + upper_end) / 2, 3)
    central_lower = round((lower_start + lower_end) / 2, 3)

    # Возвращаем результат в нужном формате
    return (central_upper, central_lower)
```

**Вычисление среднего арифметического двух чисел**

```python
# Функция вычисления среднего арифметического дефаззифицированных значений двух функций принадлежности
def calculate_average(defuzzified_values):
    # Извлекаем значения из кортежа
    defuzzified_upper, defuzzified_lower = defuzzified_values
    # Вычисляем среднее арифметическое
    average = round((defuzzified_upper + defuzzified_lower) / 2, 3)
    return average
```
